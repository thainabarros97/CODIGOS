# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# passo 4: calcular os indicadores (faturamento e quantidade de produtos vendidos) 

import pandas 

# abrir base de dados
#colocar r antes do caminho do arquivo
caminho_arquivo = r"C:\Users\thaina barros\Downloads\Vendas - Dez.xlsx"
tabela = pandas.read_excel(caminho_arquivo)
 
# ver informações da base de dados pode ser print, porém display traz melhor formatada.
display(tabela)

# somar o faturamento de todos os produtos = somar coluna de valor final.
#colocar coluna entre colchetes []
faturamento = tabela ["Valor Final"].sum()


# somar a quantidade de produtos = somar coluna quantidade.
quantidade_produtos = tabela ["Quantidade"].sum()

print(faturamento)
print(quantidade_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
import pyperclip 
# passo 5: enviar informações por e-mail
#abrir nova aba
pyautogui.hotkey("ctrl","t")

#entrar no e-mail
pyautogui.write ("https://mail.google.com/mail/u/0/")
pyautogui.press ("enter")
time.sleep(8)

pyautogui.click(x=67, y=185, clicks=1)
time.sleep(6)

#para quem enviar o e-mail 
pyautogui.write("thainabarros97@gmail.com")
pyautogui.press("tab")
pyautogui.press("tab")

#qual o assunto 
pyperclip.copy("Relatório de Vendas")
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")

# qual o corpo do e-mail
#colocar o f antes do texto para informar formatar

texto = f"""
Prezados, 

Segue o relatório de vendas de hoje. 

Faturamento: R${faturamento:,.2f}
Quantidade de produtos vendidos: {quantidade_produtos:,}

Qualquer dúvida estou a disposição.
"""
pyperclip.copy(texto)
pyautogui.hotkey("ctrl", "v")

#enviar o e-mail 
pyautogui.click(x=818, y=816, clicks=1)

```
